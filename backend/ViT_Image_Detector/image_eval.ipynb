```python
from pathlib import Path
import glob
import zipfile

# 1. Retrieve the list of entry names from the ZIP file first
zip_files = glob.glob("/content/cifake/*.zip")
if zip_files:
    with zipfile.ZipFile(zip_files[0], "r") as z:
        image_names = z.namelist()
else:
    image_names = []
    print("Warning: No zip files found in /content/cifake/")

def get_class(path):
    parts = [p.lower() for p in Path(path).parts[:-1]]

    if "real" in parts:
        return 0
    if "fake" in parts:
        return 1
    return None

real_files = [n for n in image_names if get_class(n) == 0]
fake_files = [n for n in image_names if get_class(n) == 1]

print("Real images:", len(real_files))
print("AI-generated images:", len(fake_files))
```

```python
import os
import zipfile
import random

random.seed(42)

selected = (
    [(n, 0) for n in random.sample(real_files, 2500)] +
    [(n, 1) for n in random.sample(fake_files, 2500)]
)
random.shuffle(selected)

TEST_DIR = "/content/cifake/selected_test"

with zipfile.ZipFile(zip_files[0], "r") as z:
    for i, (name, label) in enumerate(selected):
        folder = "REAL" if label == 0 else "FAKE"
        dest = os.path.join(TEST_DIR, folder, f"{i}_{os.path.basename(name)}")
        os.makedirs(os.path.dirname(dest), exist_ok=True)

        with open(dest, "wb") as f:
            f.write(z.read(name))

print("Selected 5,000 images!")
```

```python
import os

SEARCH_DIRS = [
    "/content/cifake",
    "/content/drive/MyDrive/plag_models/ViT_Image_Detector",
    "/content/drive/MyDrive"
]

for base in SEARCH_DIRS:
    print(f"\nSearching: {base}")

    if not os.path.exists(base):
        print("Path does not exist")
        continue

    for root, dirs, files in os.walk(base):
        dirs[:] = [d for d in dirs if d not in [".git", "node_modules"]]

        image_files = [
            f for f in files
            if f.lower().endswith((".jpg", ".jpeg", ".png", ".webp", ".bmp"))
        ]

        if image_files:
            print(f"{root} — {len(image_files)} images")
```

```python
from pathlib import Path
from PIL import Image
import torch
import time
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Set this to your actual 5,000-image folder
TEST_DIR = Path("/content/cifake/selected_test")

assert "model" in globals(), "Your image model is not loaded."
assert "image_processor" in globals(), "Your image processor is not loaded."

extensions = {".jpg", ".jpeg", ".png", ".webp", ".bmp"}

samples = []
for folder, label in [("REAL", 0), ("FAKE", 1)]:
    folder_path = TEST_DIR / folder
    if folder_path.exists():
        samples.extend(
            (p, label)
            for p in folder_path.rglob("*")
            if p.is_file() and p.suffix.lower() in extensions
        )

print("Images found:", len(samples))
print("Real:", sum(y == 0 for _, y in samples))
print("AI-generated:", sum(y == 1 for _, y in samples))

assert len(samples) > 0, f"No images found in {TEST_DIR}"

device = next(model.parameters()).device
model.eval()

y_true, y_pred = [], []
failed = 0
batch_size = 16
start = time.time()

for i in range(0, len(samples), batch_size):
    batch = samples[i:i + batch_size]
    images, labels = [], []

    for path, label in batch:
        try:
            with Image.open(path) as im:
                images.append(im.convert("RGB"))
            labels.append(label)
        except Exception:
            failed += 1

    if not images:
        continue

    inputs = image_processor(images=images, return_tensors="pt")
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.inference_mode():
        predictions = model(**inputs).logits.argmax(dim=-1).cpu().tolist()

    y_true.extend(labels)
    y_pred.extend(predictions)

print(f"\nEvaluation time: {time.time() - start:.1f} seconds")
print("Evaluated:", len(y_true))
print("Unreadable:", failed)

if y_true:
    print(f"Accuracy: {accuracy_score(y_true, y_pred) * 100:.2f}%")
    print("\nClassification report:")
    print(classification_report(
        y_true, y_pred,
        labels=[0, 1],
        target_names=["Real", "AI-generated"],
        digits=4,
        zero_division=0
    ))
    print("Confusion matrix:")
    print(confusion_matrix(y_true, y_pred, labels=[0, 1]))
```
